```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
hmcts = pd.read_csv('HMCTS_MI_output_accessible_Oct_26.csv')
```

```python
hmcts.info()
```

```python
hmcts.fillna(value=0,inplace=True)
```

```python
hmcts.head(10)
```

```python
fig,ax = plt.subplots(figsize=(9,5))
sns.barplot(data=hmcts,x='Civil Claims received',y='Month',ax=ax,palette='Set1',orient='h')
plt.title('Civil Claims Received on Jul 25 - Aug 26')
plt.show()
```
