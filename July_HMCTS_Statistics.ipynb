```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
hmcts26 = pd.read_csv('HMCTS_MI_output_accessible_Aug_26_V3.csv')
```

```python
hmcts26.info()
```

```python
claims = sns.barplot(data=hmcts26,x='Month',y='Percentage of Small claims that settled at mediation',palette='Set1')
claims.set(title='Percentage of Small Claims')
plt.xticks(rotation=80)
```

```python
small_claims = sns.lmplot(data=hmcts26,x='Small claims mediations',y='Small claims settled',hue='Month')
small_claims.set(title='Small claims that settled at mediation')
```

```python
famlaw1 = sns.lmplot(data=hmcts26,x='Private Family Law Receipts',y='Private Family Law Disposals',hue='Month')
famlaw1.set(title='Private Family Law 2025-2026')
```

```python
famlaw2 = sns.lmplot(data=hmcts26,x='Public Family Law Receipts',y='Public Family Law Disposals',hue='Month')
famlaw2.set(title='Public Family Law 2025-2026')
```

```python
hmcts26.plot(x='Month',y='Public Family Law Open Caseload',marker='o',color='navy',legend=False)
plt.title('Public Family Law Open Caseload Jun-26-Jul-26')
plt.ylabel('Frequency')
plt.show()
```

```python
employlaw = sns.lmplot(data=hmcts26,x='Employment Tribunal Receipts - Single and Lead  Multiple  Claims',y='Employment Tribunal Disposals  - Single and Lead Multiple Claims',hue='Month')
employlaw.set(title='Employment Tribunal Law',xlabel='Reciepts',ylabel='Disposals')
plt.show()
```

```python
emplaw2= sns.barplot(data=hmcts26, x='Month',y='Employment Tribunal Open Caseload  - Single and Lead Multiple Claims',palette='Set1')
emplaw2.set(title='Employment Tribunal Open Caseload Jun 25- Jul 26',ylabel='Single and Lead Multiple Claims')
plt.xticks(rotation=80)
plt.show()
```

This bar graph is increasing the number of single and lead multiple claims from July 2025 to July 26 which is one year difference.

```python
conorders = sns.lmplot(data=hmcts26,x='Divorce - Conditional Orders Pronounced - from joint submissions',y='Divorce - Conditional Orders Pronounced - from sole submissions',hue='Month')
conorders.set(title='Divorce Conditinal Orders Pronounced Jun 25- Jul 26',xlabel='Joint Submissions',ylabel='Sole Submissions')
plt.show()
```

```python
iachambers = sns.barplot(data=hmcts26,x='Month',y='Immigration and Asylum Chamber timeliness - average weeks from  Receipt to Disposal',palette='Set1')
iachambers.set(title='Immigration and Asylum Chamber timeliness Jun 25-Jul 26',ylabel='Average weeks from Receipt and Diposal')
plt.xticks(rotation=80)
plt.show()
```

```python
sns.barplot(data=hmcts26,x='Probate - Cases that are stopped or new cases where HMCTS is waiting for documentation',y='Month',hue='Month')
plt.title('Probate - Cases that are stopped or new cases where HMCTS is waiting for documentation')
plt.xlabel('Cases')
plt.show()
```
